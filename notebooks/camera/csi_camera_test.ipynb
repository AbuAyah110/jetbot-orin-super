# CSI camera test (CAM0 / IMX219)

Live preview from the Pi Camera v2 on **CAM0** via `nvarguscamerasrc`.
This does **not** drive motors and does **not** run collision/road models.

```python
import traitlets
import ipywidgets.widgets as widgets
from IPython.display import display
from jetbot import Camera, bgr8_to_jpeg

image = widgets.Image(format='jpeg', width=300, height=300)
display(image)

camera = Camera.instance(width=224, height=224, fps=30)
camera_link = traitlets.dlink((camera, 'value'), (image, 'value'), transform=bgr8_to_jpeg)
print('camera ok', camera.value.shape)
```

When finished, stop the capture so Argus can be reused:

```python
camera_link.unlink()
camera.stop()
print('stopped')
```
